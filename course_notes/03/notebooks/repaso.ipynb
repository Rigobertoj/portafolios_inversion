```python
df.loc["precio_t=6"] = [450.50, 210.30, 121.50, 203.40]

df.loc["allocation_t=6"] = df.loc["no_stock"] * df.loc["precio_t=6"]

df
```

```python
df.loc["pesos_t=6"] = df.loc["allocation_t=6"] / sum(df.loc["allocation_t=6"])

df
```
