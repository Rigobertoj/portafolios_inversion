# Portafolio de inversión

## Tesis de inversión

Nuestra tesis de inversión se fundamenta en la coyuntura económica actual y en el perfil del cliente descrito en las secciones previas. Bajo esta lógica, el portafolio no se construye desde una sola lectura del mercado, sino desde una combinación de enfoques que permitan capturar distintas fuentes de retorno, sin perder coherencia con el entorno económico, el horizonte de inversión y los objetivos previamente definidos.

En términos generales, la economía de Estados Unidos mantiene una expansión positiva, aunque con una narrativa menos limpia que la de un ciclo claramente expansivo. El PIB real creció 2.0% anualizado en el primer trimestre de 2026, por lo que no es correcto sostener una lectura de recesión confirmada. Sin embargo, la persistencia de inflación, la sensibilidad a tasas de interés y la incertidumbre sobre el ciclo económico obligan a construir una cartera que no dependa de un solo escenario macroeconómico. La Fed también ha señalado que la actividad económica continúa expandiéndose, aunque la inflación permanece elevada, en parte por el incremento reciente de precios globales de energía.

De tal manera que la tesis del portafolio no busca apostar de forma agresiva a una sola narrativa, sino construir una cartera capaz de responder razonablemente a distintos desenlaces: un entorno donde el mercado siga premiando crecimiento, un entorno donde se valore más la generación actual de flujo, o un entorno donde la resiliencia financiera y operativa sea más importante que la expansión esperada.

Con esto en mente, las tres filosofías seleccionadas fueron:

- **Value**
- **Growth**
- **Quality**

Cada una de estas filosofías permite observar una dimensión distinta del atractivo de una compañía. **Value** ayuda a evaluar si el mercado está castigando demasiado el precio de un activo. **Growth** permite analizar si una empresa tiene capacidad real de expandir su valor económico futuro. **Quality** busca determinar si la compañía tiene una estructura de negocio suficientemente sólida como para sostener rentabilidad superior, generación de efectivo y resiliencia a través del ciclo.

La intención no es que estas tres líneas funcionen como categorías aisladas, sino como lentes de análisis. Es decir, cada security incluida en la cartera debe poder justificar su pertenencia desde una narrativa clara: porque ofrece un descuento razonable frente a su valor intrínseco, porque tiene capacidad de crecimiento superior, o porque posee una calidad financiera y competitiva que permite reducir fragilidad dentro del portafolio.

En ese sentido, el objetivo del research no es simplemente seleccionar acciones “baratas”, “buenas” o “de crecimiento”, sino construir una argumentación metodológica y narrativa que permita explicar por qué cada activo tiene sentido dentro de la cartera.



---

### Value

La filosofía **Value** parte de la idea de que el precio de mercado no siempre refleja correctamente el valor económico de una compañía. En ciertos momentos, el mercado puede castigar a una empresa por factores temporales, incertidumbre sectorial, presión macroeconómica, resultados de corto plazo o simplemente por falta de atención. Bajo esta lógica, una inversión Value busca identificar situaciones donde existe una diferencia razonablemente defendible entre el precio actual del activo y su valor intrínseco.

De forma intuitiva, Value puede entenderse como la posibilidad de comprar un dólar por cincuenta centavos. Sin embargo, esta idea debe tratarse con cuidado, porque una empresa no clasifica como Value únicamente porque cotiza con múltiplos bajos. Una compañía puede estar barata porque el mercado está siendo ineficiente, pero también puede estar barata porque su negocio se está deteriorando, porque su balance no es sostenible o porque su flujo de efectivo no tiene la calidad suficiente.

Por tanto, dentro de esta tesis, Value se entiende como una filosofía que busca compañías con descuento frente a su valor intrínseco y frente a comparables relevantes, siempre y cuando exista evidencia de generación de flujo, balance sobrevivible, ausencia de deterioro estructural irreversible y una ruta razonable para que el mercado cierre total o parcialmente la brecha entre precio y valor.

La manera en la que se introduce esta metodología sigue una estructura de análisis en donde primero se identifica si existe una narrativa cualitativa razonable para estudiar la compañía. Es decir, si el activo parece castigado por razones que podrían ser temporales, cíclicas o exageradas. Posteriormente, esa intuición se contrasta con métricas de valuación y calidad financiera que permiten evaluar si la aparente baratura realmente tiene sustento.

Entre las métricas más relevantes se consideran múltiplos como **P/E, P/B, P/S, EV/EBITDA, EV/EBIT, FCF Yield y shareholder yield**, además del descuento frente a comparables, descuento frente a su propia historia, margen de seguridad, flujo libre de efectivo, ROIC, deuda, cobertura de intereses y calidad de utilidades.

Lo importante es que estas métricas no se interpretan de forma aislada. Un múltiplo bajo solo tiene sentido cuando se compara contra empresas con modelos de negocio, estructura de capital, márgenes, geografía y ciclo económico razonablemente similares. De tal manera que el análisis Value no busca encontrar acciones “baratas” en términos superficiales, sino compañías donde el descuento parezca excesivo en relación con su capacidad real de generar valor.

Bajo esta lógica, Value aporta al portafolio una fuente de retorno asociada a la normalización del precio, al re-rating de múltiplos, a la recuperación de expectativas o al reconocimiento gradual de valor por parte del mercado. Su principal riesgo es caer en una **Value Trap**, es decir, comprar una empresa aparentemente barata que en realidad merece cotizar con descuento por deterioro financiero, operativo o competitivo.

---

### Growth

La filosofía **Growth** parte de una pregunta distinta: no se enfoca principalmente en cuánto vale hoy la compañía respecto a sus activos o utilidades actuales, sino en cuánto puede crecer su valor económico futuro. Una empresa Growth no es simplemente una empresa que aumenta ventas. Es una compañía cuyo valor depende principalmente de su capacidad para expandir ingresos, utilidades, flujos de efectivo y valor intrínseco a una tasa superior a la del mercado, su industria o la economía.

De tal manera que Growth debe entenderse como una tesis sobre el futuro de la compañía. La pregunta central no es únicamente si la empresa crece, sino si ese crecimiento es real, sostenible, rentable, financiable y capturable por el accionista. Esto es importante porque una empresa puede crecer en tamaño y aun así destruir valor si necesita demasiada reinversión, si diluye constantemente a sus accionistas, si no logra escalar márgenes o si su crecimiento depende de adquisiciones que no generan retornos superiores al costo de capital.

Por tanto, dentro de esta tesis, una compañía Growth se considera atractiva cuando sus oportunidades futuras de crecimiento son amplias, rentables, financiables y defendibles frente a la competencia. Además, el precio actual no debe descontar una ejecución imposible, porque una gran empresa puede convertirse en una mala inversión si el mercado ya incorporó expectativas demasiado optimistas.

La forma de analizar Growth dentro del portafolio busca conectar la narrativa de expansión con evidencia financiera concreta. Primero se identifica el motor de crecimiento: nuevos clientes, mayor volumen, precios, expansión geográfica, nuevos productos, incremento en uso por cliente, penetración de mercado o crecimiento del mercado direccionable. Después se evalúa si ese crecimiento puede traducirse eventualmente en margen, utilidad operativa, flujo libre y retorno sobre capital.

Entre las métricas más importantes se consideran **crecimiento de ingresos, CAGR de ventas, crecimiento orgánico, crecimiento por acción, crecimiento de utilidad bruta, expansión de márgenes, CFO growth, FCF growth, margen bruto, FCF margin, ROIC, ROIC incremental, TAM/SAM/SOM, dilución, SBC/Revenue, cash runway, EV/Sales, EV/Gross Profit, EV/EBITDA forward, P/E forward, PEG y reverse DCF**.

El punto clave es que Growth no debe confundirse con una simple expectativa optimista. Una compañía puede crecer, pero si ese crecimiento no se convierte en flujo, no mejora márgenes, no genera retornos superiores al WACC o no se captura por acción, entonces el crecimiento no necesariamente crea valor. Por eso, dentro del portafolio, Growth se utiliza como una exposición a expansión estructural, pero siempre bajo una lectura crítica de valuación, expectativas implícitas y riesgo de ejecución.

Bajo esta lógica, Growth aporta al portafolio una fuente de retorno asociada al crecimiento futuro del valor intrínseco. Su principal riesgo es caer en una **Growth Trap**, es decir, seleccionar compañías que crecen en ventas pero que no logran convertir ese crecimiento en rentabilidad, flujo libre o valor real para el accionista.

---

### Quality

La filosofía **Quality** funciona como el eje de estabilidad fundamental del portafolio. Mientras Value se enfoca en el descuento y Growth en la expansión futura, Quality se enfoca en la capacidad de una empresa para sostener rentabilidad superior con menor fragilidad financiera y operativa.

Una compañía no debe clasificarse como Quality solamente porque tiene buena marca, baja volatilidad o reputación defensiva. En un proceso serio de research, Quality implica que existe evidencia de que el negocio genera retornos superiores sobre capital, convierte utilidades en efectivo, mantiene un balance resistente, reporta resultados confiables y posee una ventaja competitiva que protege su rentabilidad frente al paso del tiempo.

De tal manera que Quality no es simplemente “comprar empresas buenas”. La calidad solo tiene valor como tesis de inversión si puede observarse en los números y si además puede comprarse a un precio razonable. Una empresa puede ser excelente operativamente, pero si el precio ya descuenta una ejecución perfecta, el retorno esperado puede ser bajo o incluso poco atractivo.

La evaluación Quality se centra en identificar compañías capaces de sostener retornos superiores al costo de capital, con márgenes estables, flujo de efectivo recurrente, bajo riesgo financiero, buena asignación de capital y un moat defendible. Por tanto, esta filosofía permite evaluar no solo qué tan rentable es una empresa, sino qué tan probable es que esa rentabilidad se mantenga en el tiempo.

Entre las métricas más relevantes se consideran **ROIC, ROE, ROA, ROIC-WACC, margen EBIT, margen bruto, estabilidad de EPS, estabilidad de márgenes, CFO/NI, FCF conversion, accruals ratio, Net Debt/EBITDA, interest coverage, FCF/Debt, market share, pricing power, ROIC incremental y disciplina de capital allocation**.

La función de Quality dentro del portafolio es doble. Primero, ayuda a reducir la fragilidad de la cartera al favorecer empresas con balances sólidos, generación de efectivo y ventajas competitivas sostenibles. Segundo, funciona como un puente entre Value y Growth, porque una empresa Value es más atractiva si además tiene calidad financiera, y una empresa Growth se vuelve más defendible si su crecimiento empieza a transformarse en márgenes, flujo libre y retornos superiores sobre capital.

Bajo esta lógica, Quality aporta al portafolio una fuente de retorno asociada a la resiliencia, la persistencia de resultados y la protección frente a deterioro permanente de capital. Su principal riesgo es caer en una **Quality Trap**, es decir, pagar demasiado por una empresa de alta calidad o asumir que la calidad pasada necesariamente se mantendrá sin deterioro competitivo, financiero o de valuación.

---

### Integración de las tres filosofías

La combinación de **Value, Growth y Quality** permite construir una tesis de inversión más balanceada. Cada filosofía responde a una pregunta distinta, pero las tres buscan resolver el mismo problema: determinar si una compañía tiene sentido dentro de la cartera considerando su precio, su capacidad de crecimiento, su calidad financiera y su papel dentro del entorno económico actual.

**Value** aporta margen de seguridad y exposición a compañías cuyo precio podría normalizarse. **Growth** aporta exposición a compañías con capacidad de expandir su valor intrínseco futuro. **Quality** aporta estabilidad, resiliencia y menor fragilidad ante escenarios de desaceleración, inflación persistente o condiciones financieras menos favorables.

De tal manera que el portafolio no queda condicionado a una sola narrativa macroeconómica. Si el mercado favorece crecimiento estructural, Growth puede aportar mayor potencial de apreciación. Si el mercado vuelve a premiar flujos actuales, valuaciones razonables y compañías castigadas en exceso, Value puede tener un papel relevante. Si el entorno se vuelve más incierto, Quality puede ayudar a sostener la cartera mediante empresas con mayor capacidad de generación de efectivo, mejor balance y ventajas competitivas más defendibles.

Por tanto, la tesis de inversión no consiste en predecir con precisión absoluta el escenario económico, sino en estructurar una cartera que tenga coherencia frente a distintos desenlaces posibles. Cada activo seleccionado debe poder explicar su pertenencia desde una de estas tres líneas: porque el mercado lo castiga demasiado, porque puede crecer más y por más tiempo de lo que se descuenta, o porque puede sostener rentabilidad superior con menor fragilidad.

En conclusión, la selección de activos se justifica cuando existe una relación clara entre narrativa, fundamentos y métricas. No se trata únicamente de elegir acciones atractivas, sino de construir una cartera donde cada security tenga una razón metodológica y conceptual para formar parte del portafolio final.

#### Backtest

```python
back_test_start = "2017-01-01"

# yield spot curve 10 years 
risk_free_rate = 0.04437

rebalance_config = RebalanceConfig(
    lookback_months  = 12,      #Se toman para cada ventana de back 12 meses
    rebalance_months = 3,       #Se revalancea cada 3 meses
    expanding_window = False,   #Ventana movil de 12 meses
    transaction_cost = 0        #Costos de transacción 
)

backtest_config = BacktestConfig(
    tickers = tickers_portfolio_core,
    initial_capital= initial_capital * variable,
    optimization_start=start_date,
    backtest_start=back_test_start,
    end=end_date,
    benchmark_ticker=benchmark.tickers[0],
    benchmark_label="S&P500 - Benchmark",
    risk_free_rate=risk_free_rate
)
```

```python
dynamic_backtest = DynamicBacktestEngine(
    config=backtest_config,
    rebalance_config=rebalance_config
)
```

```python
# Asignación maxima por activo del core
max_allocation = 0.30

# Asignación minima por activo del core
min_allocation = 0.075

# retorno minimo esperado
min_return = 0.17

n_activos = len(tickers_portfolio_core)

strategies = [
    PostModernStrategy(
        objective="minimum_semivariance",
        config=MinimumSemivarianceConfig(
            threshold=0,
            minimum_return=min_return,
            bounds=[(min_allocation, max_allocation)] * n_activos,
        ),
    ),
    
    PostModernStrategy(
        objective="maximum_omega",
        config=MaximumOmegaConfig(
            threshold=0,
            bounds=[(min_allocation, max_allocation)] * n_activos,
        ),
    ),
    
    MeanVarianceStrategy(
        objective="minimum_variance",
        config=MinimumVarianceConfig(
            risk_free_rate=risk_free_rate,
            minimum_return=min_return,
            bounds=[(min_allocation, max_allocation)] * n_activos,
        ),
    ),
    
    MeanVarianceStrategy(
        objective="maximum_sharpe",
        config=OptimizationConfig(
            risk_free_rate=risk_free_rate,
            bounds=[(min_allocation, max_allocation)] * n_activos,
        ),
    ),
]
```

```python
results_dynamic_backtest = dynamic_backtest.run(
    strategies=strategies,
    prices=assets.get_prices(),
    benchmark_prices=benchmark.get_prices()
)
```

```python
results_dynamic_backtest.evolution.round(decimals=2)
```

```python
TRIMESTRES = [
    '2017-Q1','2017-Q2','2017-Q3','2017-Q4',
    '2018-Q1','2018-Q2','2018-Q3','2018-Q4',
    '2019-Q1','2019-Q2','2019-Q3','2019-Q4',
    '2020-Q1','2020-Q2','2020-Q3','2020-Q4',
    '2021-Q1','2021-Q2','2021-Q3','2021-Q4',
    '2022-Q1','2022-Q2','2022-Q3','2022-Q4',
    '2023-Q1','2023-Q2','2023-Q3','2023-Q4',
    '2024-Q1','2024-Q2','2024-Q3','2024-Q4',
    '2025-Q1','2025-Q2','2025-Q3','2025-Q4',
    '2026-Q1','2026-Q2',
]

TIIE = [
    6.25,6.75,7.00,7.25,
    7.50,7.75,7.75,8.25,
    8.25,8.25,7.75,7.25,
    6.50,5.00,4.50,4.25,
    4.00,4.25,4.75,5.50,
    6.50,7.75,9.25,10.50,
    11.25,11.25,11.25,11.25,
    11.00,10.75,10.25,10.00,
    9.00,8.50,7.75,7.25,
    7.00,6.50,
]

CETES28 = [
    6.32,6.82,7.05,7.30,
    7.55,7.80,7.81,8.33,
    8.28,8.26,7.79,7.27,
    6.53,5.03,4.52,4.27,
    4.01,4.26,4.80,5.56,
    6.56,7.80,9.34,10.55,
    11.29,11.31,11.33,11.29,
    11.04,10.79,10.29,10.03,
    9.03,8.53,7.79,7.27,
    7.01,6.49,
]

CETES364 = [
    6.95,7.35,7.55,7.80,
    8.05,8.30,8.30,8.90,
    8.80,8.75,8.25,7.75,
    7.00,5.40,4.85,4.55,
    4.30,4.60,5.20,6.10,
    7.20,8.50,10.05,11.00,
    11.65,11.68,11.70,11.60,
    11.30,11.05,10.55,10.28,
    9.25,8.80,8.05,7.55,
    7.25,7.00,
]

BONO10 = [
    7.79,7.12,7.11,7.74,
    7.84,7.98,8.10,9.01,
    8.26,8.20,7.54,7.00,
    8.05,6.48,6.13,5.97,
    6.98,7.80,7.45,7.72,
    8.20,9.36,10.03,10.18,
    9.11,9.10,10.24,9.77,
    9.37,10.36,9.87,9.90,
    9.46,9.65,9.37,9.37,
    9.19,9.07,
]

bullet_params = {
    "isr_anual": 0.0090,
    "dur_10a": 7.3,
    "umbral_alto": 8.50,
    "umbral_medio": 7.00,
    "max_peso_bono": 0.85,
    "take_profit": 0.10,
    "cetes28_minimo": 5.00,
}
```

```python
def run_bullet_timing_daily(
    capital=1,
    trimestres=TRIMESTRES,
    tiie=TIIE,
    cetes28=CETES28,
    cetes364=CETES364,
    bono10=BONO10,
    params=bullet_params,
    start_date="2017-01-01",
):
    dates_q = pd.PeriodIndex(trimestres, freq="Q").to_timestamp(how="end").normalize()
    start_date = pd.Timestamp(start_date)

    rows = []
    events = []

    isr_q = params["isr_anual"] / 4
    dur_10a = params["dur_10a"]

    cap_bullet = capital
    cap_cetes = capital

    tc10 = 0.0
    w10 = 0.0
    wc364 = 1.0

    rows.append({
        "date": start_date,
        "bullet_timing": cap_bullet,
        "cetes_28": cap_cetes,
        "daily_return_bullet": 0.0,
        "daily_return_cetes": 0.0,
        "w_bono10": w10,
        "w_cetes364": wc364,
        "tiie": tiie[0],
        "cetes28": cetes28[0],
        "cetes364": cetes364[0],
        "bono10": bono10[0],
        "event": "",
    })

    period_starts = [start_date] + list(dates_q[:-1])
    period_ends = list(dates_q)

    for i, (period_start, period_end) in enumerate(zip(period_starts, period_ends)):
        tiie_i = tiie[i]
        c28 = cetes28[i]
        c364 = cetes364[i]
        b10 = bono10[i]

        event = ""

        if c28 < params["cetes28_minimo"] and w10 > 0:
            w10 = 0.0
            wc364 = 1.0
            event = "salida_tasa_baja"

        elif c28 >= params["cetes28_minimo"]:
            if tiie_i >= params["umbral_alto"] and w10 < params["max_peso_bono"]:
                w10 = params["max_peso_bono"]
                wc364 = 1.0 - w10
                tc10 = b10
                event = "compra"

            elif tiie_i >= params["umbral_medio"] and w10 < 0.45:
                w10 = min(0.50, params["max_peso_bono"])
                wc364 = 1.0 - w10
                tc10 = b10
                event = "compra"

            elif tiie_i < params["umbral_medio"] and w10 > 0:
                w10 = 0.0
                wc364 = 1.0
                event = "salida_tiie_baja"

        ganancia_extra = 0.0
        if w10 > 0 and tc10 > 0:
            price_gain = -dur_10a * (b10 - tc10) / 100
            if price_gain >= params["take_profit"]:
                ganancia_extra = w10 * price_gain
                tc10 = b10
                event = "take_profit"

        r_q_bullet = (
            wc364 * (c364 / 100 / 4)
            + w10 * (tc10 / 100 / 4 if tc10 else 0)
            + ganancia_extra
            - isr_q
        )

        r_q_cetes = c28 / 100 / 4 - isr_q

        days = pd.bdate_range(period_start, period_end)

        if i == 0:
            days = days[days > start_date]
        else:
            days = days[days > period_start]

        if len(days) == 0:
            continue

        r_d_bullet = (1 + r_q_bullet) ** (1 / len(days)) - 1
        r_d_cetes = (1 + r_q_cetes) ** (1 / len(days)) - 1

        for j, day in enumerate(days):
            cap_bullet *= (1 + r_d_bullet)
            cap_cetes *= (1 + r_d_cetes)

            rows.append({
                "date": day,
                "bullet_timing": cap_bullet,
                "cetes_28": cap_cetes,
                "daily_return_bullet": r_d_bullet,
                "daily_return_cetes": r_d_cetes,
                "w_bono10": w10,
                "w_cetes364": wc364,
                "tiie": tiie_i,
                "cetes28": c28,
                "cetes364": c364,
                "bono10": b10,
                "event": event if j == 0 else "",
            })

        if event:
            events.append({"date": days[0], "event": event})

    evolution = pd.DataFrame(rows).set_index("date")
    events = pd.DataFrame(events)

    return evolution, events

```

```python
def fixed_income_metrics(evolution, capital, params=bullet_params):
    returns = evolution["daily_return_bullet"]
    benchmark_returns = evolution["daily_return_cetes"]

    years = (evolution.index[-1] - evolution.index[0]).days / 365.25

    final_capital = evolution["bullet_timing"].iloc[-1]
    final_benchmark = evolution["cetes_28"].iloc[-1]

    cumulative = evolution["bullet_timing"]
    drawdown = cumulative / cumulative.cummax() - 1

    metrics = pd.Series({
        "capital_inicial": capital,
        "capital_final_bullet": final_capital,
        "capital_final_cetes28": final_benchmark,
        "rendimiento_total_bullet": final_capital / capital - 1,
        "rendimiento_total_cetes28": final_benchmark / capital - 1,
        "alpha_total": final_capital / final_benchmark - 1,
        "cagr_bullet": (final_capital / capital) ** (1 / years) - 1,
        "cagr_cetes28": (final_benchmark / capital) ** (1 / years) - 1,
        "volatilidad_anualizada": returns.std() * np.sqrt(252),
        "max_drawdown": drawdown.min(),
        "peso_promedio_bono10": evolution["w_bono10"].mean(),
        "peso_max_bono10": evolution["w_bono10"].max(),
        "duracion_promedio_aprox": (evolution["w_bono10"] * params["dur_10a"]).mean(),
        "dias_con_bono10": (evolution["w_bono10"] > 0).mean(),
        "carry_promedio_aprox": (
            evolution["w_bono10"] * evolution["bono10"]
            + evolution["w_cetes364"] * evolution["cetes364"]
        ).mean() / 100,
    })

    return metrics
```
