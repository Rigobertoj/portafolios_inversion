```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(columns=["ACWI", "IUII", "SPLG"])

df.loc["pesos_objetivo"] = [0.30, 0.30, 0.40]

df.loc["precio_t=0"] = [91.25, 20.21, 51.14]

df
```

```python
df.loc["num_stock"] = np.floor((200_000 * df.loc["pesos_objetivo"]) / df.loc["precio_t=0"])
df
```

```python
df2 = pd.DataFrame(columns=["ACWI", "IUII", "SPLG"])

df2.loc["precio_t=6m"] = [87.45, 20.17, 77.15]

df2.loc["valor_posicion"] = df.loc["num_stock"] * df2.loc["precio_t=6m"]

df2.loc["pesos_actuales"] = df2.loc["valor_posicion"] / sum(df2.loc["valor_posicion"])

df2
```

```python
sum(df2.loc["valor_posicion"])
```

```python
df2.loc["Revalanceo C/V"] = np.floor((df.loc["pesos_objetivo"] - df2.loc["pesos_actuales"]) * sum(df2.loc["valor_posicion"]) /df2.loc["precio_t=6m"])

df2
```

```python
# Comprobación rapida
(
    df.loc["num_stock"] + 
    df2.loc["Revalanceo C/V"]
) * df2.loc["precio_t=6m"]/sum(df2.loc["valor_posicion"])
```
